```python
filename = "my_data.txt"

with open(filename, "r") as file:
    lines = file.readlines()

print("".join(lines))
```

```python
# File statistics
print("Number of lines:", len(lines))
print("Number of words:", sum(len(line.split()) for line in lines))
```
